```python
import sqlite3
import pandas as pd

# 1. Initialize an in-memory SQL database environment
conn = sqlite3.connect(':memory:')

# 2. Load our raw CSV files seamlessly into relational database tables
orders = pd.read_csv('../data/olist_orders_dataset.csv')
order_items = pd.read_csv('../data/olist_order_items_dataset.csv')
products = pd.read_csv('../data/olist_products_dataset.csv')
customers = pd.read_csv('../data/olist_customers_dataset.csv')

orders.to_sql('orders', conn, index=False, if_exists='replace')
order_items.to_sql('order_items', conn, index=False, if_exists='replace')
products.to_sql('products', conn, index=False, if_exists='replace')
customers.to_sql('customers', conn, index=False, if_exists='replace')

print("✅ Relational database tables generated successfully! Engine online.")
```

```python
# Query 1: Top 5 Customer States by Gross Transaction Value
query_1 = """
SELECT 
    c.customer_state AS State,
    COUNT(DISTINCT o.order_id) AS Total_Orders,
    ROUND(SUM(oi.price), 2) AS Gross_Revenue
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
JOIN customers c ON o.customer_id = c.customer_id
WHERE o.order_status = 'delivered'
GROUP BY c.customer_state
ORDER BY Gross_Revenue DESC
LIMIT 5;
"""
pd.read_sql_query(query_1, conn)
```

```python
# Query 2: Fulfillment Logistics & Delivery Backlogs by Order Status
query_2 = """
SELECT 
    order_status AS Order_Status,
    COUNT(*) AS Order_Count
FROM orders
GROUP BY order_status
ORDER BY Order_Count DESC;
"""
pd.read_sql_query(query_2, conn)
```
